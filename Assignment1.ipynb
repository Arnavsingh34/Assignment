FOR LOOP AND RANGE FUNCTIONS




```python
# Q.1 Write a program to print numbers from 1 to 10 using a for loop.

for i in range(1, 11):
    print(i)
```

```python
# Q.2 Write a program to print all even numbers between 1 and 50 using range().
for i in range(1, 51):
    if i % 2 == 0:
        print(i)
```

```python
# Q.3 Write a program to print the multiplication table of a given number  using a for loop.
number = 5  # You can change this value to print the multiplication table of any number
for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
```

```python
# Q.4 Write a program to calculate the sum of all numbers from 1 to 100 using a for loop.
total = 0
for i in range(1, 101):
    total += i

print("sum=", total)
```

```python
# Q.5 Write a program to print numbers from 10 to 1 in reverse order using range()
for i in range(10, 0, -1):
    print(i)

```

 WHILE LOOP

```python
# Q.6 Write a program to print numbers from 1 to 10 using a while loop.
i = 1
while i <= 10:
    print(i)
    i += 1
```

```python
# Q.7 Write a program to find the factorial of a number using a while loop.
num = int(input("Enter a number: "))

factorial = 1
i = 1

while i <= num:
    factorial *= i
    i += 1

print("Factorial =", factorial)
```

```python
# Q.9 Write a program to count how many digits are in a given number using a while loop.
num = int(input("Enter a number:"))
count = 0
while num != 0:
    num = num // 10
    count += 1
print("number of digits=", count)
```

```python
# Q.10 Write a program to print the Fibonacci series up to n terms using a while loop.
n = int(input("Enter the number of  terms:"))

a = 0
b = 1
i = 1

while i <= n:
    print(a, end=" ")
    c = a + b
    a = b
    b = c
    i += 1
```

```python
# Q.15 Write a function count_vowels(word) that returns the number of vowels in a given word.
def count_vowels(word):
    vowels = "aeiouAEIOU"
    count = 0
    for char in word:
        if char in vowels:
            count = count+ 1

            return count
    
word = input("Enter a word:")

print("number of vowels=", count_vowels(word))
```
