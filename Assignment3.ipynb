```python
# given array
arr = [23, 43, 13, 27, 37]

# hash table size
size = 10

# create empty hash tble
hash_table = [None] * size

# insert every number
for ele in arr:

    # calculate hash index
    index = ele % size

    # if index is already occupied
    while hash_table[index] is not None:

        # move to next index
        index = (index + 1) % size

    # store the number at empty iindex
    hash_table[index] = ele

# print the hash table
print(hash_table)
```
